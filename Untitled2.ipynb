```python
# STEP 1: Install libraries
!pip -q install transformers sentencepiece torch matplotlib seaborn

import torch
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
from IPython.display import display, Markdown

# STEP 2: Load the pretrained Transformer
MODEL_NAME = "sshleifer/distilbart-cnn-12-6"

device = "cuda" if torch.cuda.is_available() else "cpu"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME)
model = model.to(device)
model.eval()

print("Model:", MODEL_NAME)
print("Device:", device)

# STEP 3: Input research text
input_text = """
Artificial intelligence is increasingly used in education.
Machine learning systems can personalize learning by analyzing
student performance. Recent research shows that personalized
feedback can improve student engagement and learning outcomes.
However, these systems require high-quality data and careful
evaluation. Privacy, bias, and computational costs remain
important challenges. Future research should investigate
reliable evaluation methods and responsible AI deployment.
"""

display(Markdown("## 1. Input Research Text"))
print(input_text)

# STEP 4: Tokenization
inputs = tokenizer(
    input_text,
    return_tensors="pt",
    truncation=True,
    max_length=512
).to(device)

token_ids = inputs["input_ids"][0]
tokens = tokenizer.convert_ids_to_tokens(
    token_ids.tolist()
)

display(Markdown("## 2. Tokenization"))
print("Number of tokens:", len(tokens))
print("Token IDs:", token_ids.tolist())
print("Tokens:")
print(tokens)

# Visualize token IDs
plt.figure(figsize=(12, 3))
plt.bar(range(min(25, len(tokens))),
        token_ids[:25].cpu().numpy())
plt.xticks(
    range(min(25, len(tokens))),
    tokens[:25],
    rotation=60,
    ha="right"
)
plt.title("Token IDs (first 25 tokens)")
plt.ylabel("Token ID")
plt.tight_layout()
plt.show()

# STEP 5: Run the Transformer and capture its layers
with torch.no_grad():
    outputs = model(
        **inputs,
        output_hidden_states=True,
        output_attentions=True,
        return_dict=True
    )

display(Markdown("## 3. Token Embeddings and Transformer Layers"))

# Encoder hidden states: embedding output + each encoder layer
hidden_states = outputs.encoder_hidden_states

if hidden_states is not None and len(hidden_states) > 0:
    print("Number of encoder hidden-state tensors:",
          len(hidden_states))
    print("Embedding output shape:",
          tuple(hidden_states[0].shape))

    for i, hidden in enumerate(hidden_states):
        label = "Embedding output" if i == 0 else f"Encoder layer {i}"
        print(f"{label}: {tuple(hidden.shape)}")

    # Compare representation changes across layers
    layer_changes = []

    for i in range(1, len(hidden_states)):
        if i < len(hidden_states):
            previous = hidden_states[i - 1][0]
            current = hidden_states[i][0]

            change = torch.norm(current - previous, dim=-1).mean().item()
            layer_changes.append(change)

    if len(layer_changes) > 0:
        plt.figure(figsize=(8, 4))
        plt.plot(
            range(1, len(hidden_states)),
            layer_changes,
            marker="o"
        )
        plt.xlabel("Transformer Encoder Layer")
        plt.ylabel("Mean Hidden-State Change")
        plt.title("How Token Representations Change Across Layers")
        plt.grid(True)
        plt.tight_layout()
        plt.show()
else:
    print("No hidden states were returned by the model.")

# STEP 6: Visualize attention between input tokens
display(Markdown("## 4. Transformer Self-Attention"))

# Last encoder layer, averaged over attention heads
if outputs.encoder_attentions is not None and len(outputs.encoder_attentions) > 0:
    attention = outputs.encoder_attentions[-1][0]
    attention = attention.mean(dim=0).cpu().numpy()

    # Display a manageable number of tokens
    n = min(20, len(tokens))

    plt.figure(figsize=(11, 8))
    sns.heatmap(
        attention[:n, :n],
        xticklabels=tokens[:n],
        yticklabels=tokens[:n],
        cmap="viridis"
    )
    plt.title("Encoder Self-Attention (Last Layer, Head Average)")
    plt.xlabel("Tokens Being Attended To")
    plt.ylabel("Query Tokens")
    plt.xticks(rotation=60, ha="right")
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()

    print(
        "Attention helps tokens incorporate contextual information "
        "from other tokens."
    )
else:
    print("No encoder attentions were returned by the model.")

# STEP 7: Generate the summary
display(Markdown("## 5. Transformer Decoder and Summary Generation"))

with torch.no_grad():
    summary_ids = model.generate(
        **inputs,
        max_new_tokens=100,
        min_new_tokens=20,
        num_beams=4,
        early_stopping=True
    )

summary = tokenizer.decode(
    summary_ids[0],
    skip_special_tokens=True
)

display(Markdown("### Final Research Paper Summary"))
display(Markdown(summary))

# STEP 8: Show the complete pipeline
display(Markdown("""
## Pipeline Completed

**Input text**
↓
**Tokenization and token IDs**
↓
**Token embeddings + positional information**
↓
**Transformer encoder layers**
↓
**Contextual representations and self-attention**
↓
**Transformer decoder generates summary tokens**
↓
**Final coherent summary**
"""))
```
